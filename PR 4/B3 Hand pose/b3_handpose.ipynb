```python
from pathlib import Path

# Search common project locations for the exported metadata
search_roots = [
    Path.home() / "OneDrive" / "Desktop",
    Path.home() / "Desktop",
    Path.home() / "Downloads",
]

matches = []

for root in search_roots:
    if root.exists():
        matches.extend(root.glob("**/models/pose/metadata.json"))

if matches:
    METADATA_PATH = matches[0]
    POSE_DIR = METADATA_PATH.parent

    print("✅ Found pose model folder:", POSE_DIR)
    print("Files:")
    for name in ["metadata.json", "model.json", "weights.bin"]:
        p = POSE_DIR / name
        print(name, ":", "Found" if p.exists() else "Missing")
else:
    print("❌ Could not find models/pose/metadata.json")
    print("Please check the exact location of your DL_PR4B folder.")
```
